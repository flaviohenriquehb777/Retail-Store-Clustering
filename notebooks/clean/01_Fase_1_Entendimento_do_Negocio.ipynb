## Problema de negócio
- Parar de comparar loja de shopping com posto de gasolina.
- Criar grupos de lojas PARECIDAS de verdade.
- Definir metas por cluster (comparando ao seu par) e não à rede toda.

## Consumidores do resultado
- Comercial (metas)
- Trade Marketing (verba de promoção)
- Supply (estoque e sortimento)
- Expansão (qual formato abrir em um ponto novo)

## 5 critérios de sucesso com testes concretos
1. **Interpretável** (critério de gerente): cada cluster tem nome de 2-3 palavras que um gerente de rede repetiria numa reunião.
2. **Acionável** (nenhum par com mesma ação): para todos os C(k,2)=10 pares de clusters, as ações em Comercial/Trade/Supply são diferentes.
3. **Estável** (não depende de random_state):
   - Shuffle linhas com seed 42 → ARI vs labels original ≥ 0,90.
   - 10 random_states diferentes → matriz de ARI ≥ 0,85 (média ≥ 0,90).
   - Bootstrap 80% de amostra, 10 rodadas → min ARI ≥ 0,60, média ≥ 0,75.
4. **Melhor que baseline (tipo_loja / sortimento)**:
   - Concordância ARI / NMI vs tipo_loja & sortimento.
   - Pseudo R² ANOVA em pelo menos 4 métricas de negócio (VM, clientes, ticket, volatilidade, sazonalidade).
5. **Quantificado em R$ (não em silhueta)**:
   - Gap = (P75 do cluster − atual) × 365 dias.
   - 3 cenários de recuperação do gap: 30% / 50% / 70%.
